```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Loading the Titanic dataset
df = pd.read_csv("C:/Users/Hitesh Girase/OneDrive/Desktop/Ds_Assignment_Datasets/Titanic-Dataset.csv")
```

```python
# Plotting the distribution of ticket prices ('fare') using a histogram
plt.figure(figsize=(10, 6))
sns.histplot(df['Fare'], kde=True, bins=30, color='blue')
```

```python
# Customizing the plot

plt.title('Distribution of Ticket Prices (Fare) for Titanic Passengers', fontsize=16)
plt.xlabel('Fare', fontsize=14)
plt.ylabel('Frequency', fontsize=14)
plt.grid(True)
```

```python
# number of passenger per class # counterplot
sns.countplot(x='Pclass', data=df)
plt.show()
```

```python
# average fare by class # barplot
sns.barplot(x='Pclass', y='Fare', data=df)
plt.show()
```
