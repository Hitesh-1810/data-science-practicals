Project Title: Document Preprocessing and Term Frequency-Inverse Document Frequency (TF-IDF) Calculation EDA NLTK

```python
# Importing Libraries

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Downloading necessary NLTK data files

nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# Sample Document

document = "Text Analytics is a crucial step in analyzing unstructured data. It involves tokenization, lemmatization, and stop words removal."
```

```python
# Tokenization

tokens = word_tokenize(document)
print("Tokens:", tokens)
```

```python
# POS Tagging

pos_tags = nltk.pos_tag(tokens)
print("POS Tags:", pos_tags)
```

```python
# Stop Words Removal

stop_words = set(stopwords.words('english'))
```

```python
print(stop_words)
```

```python
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
print("Filtered Tokens:", filtered_tokens)
```

```python
# Stemming

stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]
print("Stemmed Words:", stemmed_words)
```

```python
# Lemmatization

lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print("Lemmatized Words:", lemmatized_words)
```

```python
# TF-IDF Calculation

corpus = [document] # List of documents
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(corpus)
```

```python
# Displaying TF-IDF scores

print("TF-IDF Matrix:")
print(tfidf_matrix.toarray())
print("Feature Names (words):", vectorizer.get_feature_names_out())
```
