```python
# Histogram for Each Feature
```

```python
# Plotting histograms for each feature
df.hist(bins=10, figsize=(10, 8))
plt.suptitle('Histograms of Iris Dataset Features')
plt.show()

```

```python
# Boxplot for Each Feature
```

```python
# Plotting boxplots for each feature

plt.figure(figsize=(10, 8))
sns.boxplot(data=df.drop('species', axis=1))
plt.title('Box Plot of Features in the Iris Dataset')
plt.show()
```

```python
# Compare Distributions and Identify Outliers
```

```python
# Visualizing histograms and boxplots for each feature grouped by species

plt.figure(figsize=(12, 8))
for i, feature in enumerate(df.columns[:-1]):
    plt.subplot(2, 2, i + 1)
    sns.histplot(data=df, x=feature, hue="species", kde=True)
    plt.title(f'Distribution of {feature}')
    plt.tight_layout()
    plt.show()
```

```python
# Comparing boxplots for different species
plt.figure(figsize=(12, 8))
for i, feature in enumerate(df.columns[:-1]):
    plt.subplot(2, 2, i + 1)
    sns.boxplot(x='species', y=feature, data=df)
    plt.title(f'Box Plot of {feature} by Species')
    plt.tight_layout()
    plt.show()
```
