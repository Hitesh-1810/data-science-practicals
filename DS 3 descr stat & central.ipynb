```python
# Filter data for different species

setosa = df[df['species'] == 'Iris-setosa']
versicolor = df[df['species'] == 'Iris-versicolor']
virginica = df[df['species'] == 'Iris-virginica']
```

Calculate Basic Statistical Measures

```python
# Statistical details for Iris-setosa

setosa_stats = setosa.describe()
print("Iris-setosa Statistics:\n", setosa_stats)
```

```python
# Statistical details for Iris-versicolor

versicolor_stats = versicolor.describe()
print("\nIris-versicolor Statistics:\n", versicolor_stats)
```

```python
# Statistical details for Iris-virginica

virginica_stats = virginica.describe()
print("\nIris-virginica Statistics:\n", virginica_stats)
```

```python
# Percentile Calculation
```

```python
# Calculate specific percentiles (e.g., 10th, 90th percentile)

percentile_10_setosa = np.percentile(setosa['sepal_length'], 10)
percentile_90_setosa = np.percentile(setosa['sepal_length'], 90)
print(f"\n10th Percentile of Sepal Length (Iris-setosa): {percentile_10_setosa}")
print(f"90th Percentile of Sepal Length (Iris-setosa): {percentile_90_setosa}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Plot histogram for SepalLength of Iris-setosa

sns.histplot(setosa['sepal_length'], kde=True)
plt.title('Distribution of Sepal Length (Iris-setosa)')
plt.show()
```

```python
sns.histplot(versicolor['sepal_length'], kde=True)
plt.title('Distribution of Sepal Length (Iris-versicolor)')
plt.show()
```
