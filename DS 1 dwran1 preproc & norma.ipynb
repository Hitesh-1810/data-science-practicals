```python
# Convert 'Age' column to numeric if not already
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
```

```python
# Convert 'Survived' column to categorical
```

```python
df['Survived'] = df['Survived'].astype('category')
```

```python
# Normalize numeric columns (e.g., 'Age', 'Fare')
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])
```

```python
# Check the types after conversion
```

```python
print("Data Types After Conversion:\n", df.dtypes)
```

```python
# Turn Categorical Variables into Quantitative Variables
```

```python
# Convert 'Sex' column to numerical (0 = male, 1 = female)
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
```

```python
# Convert 'Embarked' column to numerical (using one-hot encoding)
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)
```
