Data Visualization II: Box Plot for Distribution of Age with Respect to Gender and Survival

```python
# Importing necessary libraries

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the Titanic dataset

df = pd.read_csv("C:/Users/Hitesh Girase/OneDrive/Desktop/Ds_Assignment_Datasets/Titanic-Dataset.csv")
```

```python
df.describe
```

```python
# Plotting box plot for age distribution by gender and survival status

plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Sex', y='Age', hue='Survived')
```

```python
# Adding labels and title

plt.title('Box Plot of Age Distribution by Gender and Survival Status')
plt.xlabel('Gender')
plt.ylabel('Age')
plt.legend(title='Survived', loc='upper right', labels=['Did Not Survive', 'Survived'])
```
